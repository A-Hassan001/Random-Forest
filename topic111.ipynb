````python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
````

````python
data = load_breast_cancer()

X = data.data
y = data.target

feature_names = data.feature_names

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3, stratify=y)
````

````python
data
````

````python
# Instantiate the Random Forest Classifier model
rf = RandomForestClassifier(
    n_estimators=100,   # Set the number of decision trees in the forest
    max_depth=None,    # Allow trees to grow fully until all leaves are pure or contain fewer samples than min_samples_split
    bootstrap=True,    # Enable bootstrap sampling (drawing samples with replacement to train each tree)
    criterion='entropy',
    random_state=42,   # Set a fixed seed for the random number generator to ensure reproducible results
    n_jobs=-1           # Restrict the model to run on a single CPU core during training and prediction
)

rf.fit(X_train, y_train)
````

````python
y_predict = rf.predict(X_test)
````

````python
print(f"\nRandom Forest Accuracy: {accuracy_score(y_test, y_predict):.3f}\n")
````

````python
print(f"\nConfusion Matrix :\n")
confusion_matrix(y_test, y_predict)
````

````python
print(f"\nClassification Report:\n")
print(classification_report(y_test, y_predict))
````

````python
# 5) Show how trees contribute (Model insight)

print("\nFeature importance (average across all trees):")  
# Prints a header title with a new line before it.

for name, imp in sorted(
# Begins a loop over a sorted list of pairs, unpacking each pair into 'name' and 'imp'.

    zip(feature_names, rf.feature_importances_),
    # Pairs each feature name with its corresponding importance score calculated by the Random Forest model.

    key=lambda x: x[1],
    # Tells the sorting function to sort by the second element of each pair (the importance value).

    reverse=True,
    # Sorts the items in descending order (highest importance score first).

)[:5]:
# Slices the sorted list to take only the top 5 most important features.

    print(f"{name}: {imp:.4f}")
    # Prints the feature name and its importance score formatted to 4 decimal places.
````

## Code Implementation

### 1. Import Libraries

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
```

* `numpy` is used for numerical operations.
* `load_breast_cancer` loads the built-in breast cancer dataset.

### 2. Load Dataset

```python
data = load_breast_cancer()
```

Loads the breast cancer dataset into `data`.

### 3. Separate Features and Target

```python
X = data.data
y = data.target
```

* `X` contains the **input features** used for prediction.
* `y` contains the **target/classes** that the model needs to predict.

### 4. Get Feature Names

```python
feature_names = data.feature_names
```

Stores the names of all features so we can later understand which features are important.

### 5. Split the Data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.3, stratify=y
)
```

Divides the dataset into:

* **70% training data** → used to train the model.
* **30% testing data** → used to evaluate the model.
* `random_state=42` → gives the same split every time.
* `stratify=y` → keeps the class distribution similar in both sets.

### 6. Create Random Forest Model

```python
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    bootstrap=True,
    criterion='entropy',
    random_state=42,
    n_jobs=-1
)
```

This creates the Random Forest.

* `n_estimators=100` → creates **100 decision trees**.
* `max_depth=None` → trees can grow without a specified maximum depth.
* `bootstrap=True` → each tree gets a randomly sampled training dataset.
* `criterion='entropy'` → uses entropy to decide the best splits.
* `random_state=42` → makes the results reproducible.
* `n_jobs=-1` → uses **all available CPU cores**.

### 7. Train the Model

```python
rf.fit(X_train, y_train)
```

The Random Forest learns patterns from the training data.

### 8. Make Predictions

```python
y_predict = rf.predict(X_test)
```

The trained Random Forest uses the test features to predict their classes.

### 9. Calculate Accuracy

```python
accuracy_score(y_test, y_predict)
```

Compares the **actual answers** with the **predicted answers** to calculate accuracy.

### 10. Confusion Matrix

```python
confusion_matrix(y_test, y_predict)
```

Shows how many predictions were:

* Correctly classified
* Incorrectly classified

### 11. Classification Report

```python
classification_report(y_test, y_predict)
```

Provides:

* **Precision** → how many predicted positives were actually positive.
* **Recall** → how many actual positives were correctly found.
* **F1-score** → balance between precision and recall.
* **Support** → number of actual samples in each class.

### 12. Feature Importance

```python
rf.feature_importances_
```

Shows how important each feature was for the Random Forest's predictions.

### 13. Pair Feature Names with Importance

```python
zip(feature_names, rf.feature_importances_)
```

Connects each feature name with its importance value.

Example:

```text
Feature A → 0.25
Feature B → 0.18
Feature C → 0.10
```

### 14. Sort Features

```python
sorted(..., key=lambda x: x[1], reverse=True)
```

Sorts features from **highest importance to lowest importance**.

### 15. Select Top 5

```python
[:5]
```

Takes only the **5 most important features**.

### 16. Print Results

```python
print(f"{name}: {imp:.4f}")
```

Displays the feature name and its importance score with **4 decimal places**.

### Overall Flow

**Dataset → Features & Target → Train/Test Split → Random Forest → Training → Prediction → Evaluation → Feature Importance**
